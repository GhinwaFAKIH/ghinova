## 1. Understand the task

The graph describes the current facts. The rule flags a tool for review when a measurement in the same unit exceeds its limit. The evidence fields keep the source of each premise available for review.

```python
facts = [
    ("drill press", "vibration", (9, "mm/s"), "inspection-A"),
    ("drill press", "vibration limit", (7, "mm/s"), "procedure-1"),
    ("bench grinder", "vibration", (5, "mm/s"), "inspection-B"),
    ("bench grinder", "vibration limit", (7, "mm/s"), "procedure-1"),
    ("lathe", "vibration", (9, "mm/s"), "inspection-C"),
    ("lathe", "vibration limit", (7, "cm/s"), "procedure-2"),
]

def index_facts(facts):
    return {(entity, relation): (value, source)
            for entity, relation, value, source in facts}

index = index_facts(facts)
print(index["drill press", "vibration"])
```

## 2. Try it yourself

Write a function that evaluates one tool. Return a status and both source IDs when the units match. Return a separate status for missing data and for incompatible units. Check the drill press, grinder and lathe.

## 3. Check your work

Run the reference solution after trying the exercise. Use the assertions to check your result.

```python
def evaluate_tool(entity, index):
    measurement = index.get((entity, "vibration"))
    limit = index.get((entity, "vibration limit"))
    if measurement is None or limit is None:
        return "missing data", ()
    (value, unit), measurement_source = measurement
    (maximum, limit_unit), limit_source = limit
    sources = (measurement_source, limit_source)
    if unit != limit_unit:
        return "check units", sources
    return ("review" if value > maximum else "within limit"), sources

assert evaluate_tool("drill press", index) == ("review", ("inspection-A", "procedure-1"))
assert evaluate_tool("bench grinder", index)[0] == "within limit"
assert evaluate_tool("lathe", index)[0] == "check units"
assert evaluate_tool("unknown tool", index)[0] == "missing data"
print(evaluate_tool("drill press", index))
```
