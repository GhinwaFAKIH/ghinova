## 1. Understand the task

This exercise works with sample passages and sample answers. A real RAG application would send the prompt to a language model. Checking citation IDs can catch invented references. A separate review must determine whether the cited passages actually support the claims.

```python
import re

passages = {
    "S1": "The workshop opens at 10:00 a.m. on Saturdays and Sundays.",
    "S2": "Safety glasses are required while using the drill press.",
}
question = "When does the workshop open on Sunday?"

def build_prompt(question, passages):
    context = "\n".join(f"[{key}] {value}" for key, value in passages.items())
    instructions = (
        "Answer using only the passages below. Cite each factual claim "
        "with its source ID in brackets. If evidence is insufficient, "
        "say so."
    )
    return f"{instructions}\n\nPassages\n{context}\n\nQuestion: {question}"

print(build_prompt(question, passages))
```

## 3. Check your work

Run the reference solution after trying the exercise. Use the assertions to check your result.

```python
def check_citations(answer, passages):
    cited = set(re.findall(r"\[(S\d+)\]", answer))
    return {"cited": cited, "unknown": cited - passages.keys()}

assert check_citations("It opens at 10:00 a.m. [S1]", passages)["unknown"] == set()
assert check_citations("It opens at 10:00 a.m. [S3]", passages)["unknown"] == {"S3"}
assert check_citations("It opens at noon [S1]", passages)["unknown"] == set()
print("The noon example passes the ID check. Reading S1 reveals the factual error.")
```
