## 1. Understand the task

Each triple connects a subject to an object through a named relation. A two-hop retrieval query can find entities whose own labels do not mention the original entity. The path records the relationships behind the result.

```python
triples = [
    ("drill press", "requires", "safety glasses"),
    ("safety glasses", "stored in", "tool cabinet"),
    ("soldering station", "located in", "electronics room"),
    ("tool cabinet", "located in", "workshop"),
]

def outgoing(triples, subject, relation):
    return [obj for node, edge, obj in triples if node == subject and edge == relation]

print(outgoing(triples, "drill press", "requires"))
```

## 2. Try it yourself

Write a two-hop traversal that retrieves where the equipment required by the drill press is stored. Return the two triples used as evidence. Check that an unrelated tool returns no path.

## 3. Check your work

Run the reference solution after trying the exercise. Use the assertions to check your result.

```python
def two_hop_paths(triples, start, first, second):
    paths = []
    for middle in outgoing(triples, start, first):
        for end in outgoing(triples, middle, second):
            paths.append(((start, first, middle), (middle, second, end)))
    return paths

paths = two_hop_paths(triples, "drill press", "requires", "stored in")
assert paths == [(("drill press", "requires", "safety glasses"),
                  ("safety glasses", "stored in", "tool cabinet"))]
assert two_hop_paths(triples, "soldering station", "requires", "stored in") == []
print("Evidence path:", paths[0])
```
