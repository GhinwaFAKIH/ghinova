## Part 1 — Tokenization
A tokenizer maps text to integer token IDs. A token is not necessarily a whole word.

```python
import tiktoken
enc = tiktoken.get_encoding("cl100k_base")
prompt = "Summarize this report."
ids = enc.encode(prompt)
print("Prompt:", prompt)
print("Token IDs:", ids)
print("Token pieces:", [enc.decode([i]) for i in ids])
print("Number of tokens:", len(ids))
```

```python
for sentence in ["AI is useful.", "Artificial intelligence is useful.", "ChatGPT can summarize a report."]:
    print(repr(sentence), "→", len(enc.encode(sentence)), "tokens")
```

**Try it:** Change the prompt to include emojis, French text, or punctuation. How does the token count change?

Note: `cl100k_base` is an example tokenizer; tokenization differs across models.

## Part 2 — Sentence embeddings
A sentence embedding is a vector intended to capture aspects of sentence meaning. This demonstration uses an open-source sentence-transformer, **not ChatGPT’s internal embeddings**.

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
sentences = ["The cat is sleeping on the sofa.", "A kitten is resting on the couch.", "The stock market rose today."]
vectors = model.encode(sentences)
print("Embedding shape:", vectors.shape)
print("First 8 values of sentence 1:", vectors[0][:8])
```

```python
from sentence_transformers import util
similarities = util.cos_sim(vectors, vectors)
for i in range(len(sentences)):
    for j in range(i+1, len(sentences)):
        print(f"Sentence {i+1} vs {j+1}: {similarities[i,j].item():.3f}")
```

**Try it:** Replace the sentences with three of your own. Which pair has the highest cosine similarity? Does the result match your intuition?

**Remember:** Sentence embeddings used for semantic search are different from the token embeddings used inside a generative Transformer.
