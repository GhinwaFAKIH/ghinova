Run each cell in order. You will load data, split it, train Logistic Regression, make predictions, evaluate generalization, and experiment with the model.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## 1. Load the labeled Iris dataset
The four flower measurements are features **X**. The flower species is target **y**.

```python
iris = load_iris(as_frame=True)
X = iris.data
y = iris.target
display(X.head())
print('Target names:', iris.target_names)
```

## 2. Create unseen test data
Keep 20% out of training so we can test generalization.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print('Training examples:', len(X_train))
print('Test examples:', len(X_test))
```

## 3. Train the classifier
`fit()` estimates model parameters from the training examples.

```python
model = LogisticRegression(max_iter=300)
model.fit(X_train, y_train)
```

## 4. Inference on unseen examples
`predict()` applies the learned model to held-out test features.

```python
y_pred = model.predict(X_test)
print('First predictions:', y_pred[:10])
```

## 5. Evaluate the model

```python
print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nConfusion matrix:\n', confusion_matrix(y_test, y_pred))
print('\nClassification report:\n', classification_report(y_test, y_pred, target_names=iris.target_names))
```
