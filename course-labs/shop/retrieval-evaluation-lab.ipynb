## 1. Understand the task

A hit at k means a relevant document appears among the first k results. Reciprocal rank is 1 divided by the rank of the first relevant result, or 0 if there is no hit. Average it over labeled queries to obtain MRR.

```python
labels = {
    "lab opening time": {"hours"},
    "visitor access": {"guests"},
    "eye protection": {"safety"},
}
baseline = {
    "lab opening time": ["equipment", "hours", "guests"],
    "visitor access": ["hours", "equipment", "guests"],
    "eye protection": ["safety", "hours", "equipment"],
}
revised = {
    "lab opening time": ["hours", "equipment", "guests"],
    "visitor access": ["guests", "hours", "equipment"],
    "eye protection": ["equipment", "safety", "hours"],
}

def reciprocal_rank(ranked_ids, relevant_ids):
    for position, document_id in enumerate(ranked_ids, start=1):
        if document_id in relevant_ids:
            return 1 / position
    return 0.0

print(reciprocal_rank(baseline["lab opening time"], labels["lab opening time"]))
```

## 3. Check your work

Run the reference solution after trying the exercise. Use the assertions to check your result.

```python
def evaluate(rankings, k):
    hits = [bool(set(rankings[query][:k]) & relevant) for query, relevant in labels.items()]
    reciprocal_ranks = [reciprocal_rank(rankings[query], relevant)
                        for query, relevant in labels.items()]
    return {"hit_rate": sum(hits) / len(hits),
            "mrr": sum(reciprocal_ranks) / len(reciprocal_ranks)}

assert evaluate(baseline, 1)["hit_rate"] == 1 / 3
assert evaluate(revised, 1)["hit_rate"] == 2 / 3
assert evaluate(revised, 2)["hit_rate"] == 1
assert evaluate(revised, 2)["mrr"] > evaluate(baseline, 2)["mrr"]
print("Baseline:", evaluate(baseline, 2))
print("Revised:", evaluate(revised, 2))
```
