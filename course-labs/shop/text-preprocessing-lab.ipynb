## 1. Understand the task

The example lowercases words and removes a small list of common words. In real work, keep words that affect meaning, especially negation. Tokenization choices also matter for accents, punctuation and contractions.

```python
import re
from collections import Counter

STOP_WORDS = {"a", "an", "the", "and", "is", "are"}

def tokenize(text):
    words = re.findall(r"\b[^\W\d_]+\b", text.casefold())
    return [word for word in words if word not in STOP_WORDS]

messages = ["The lab is open and the lab is quiet.", "The lab is not open."]
for message in messages:
    print(message)
    print(Counter(tokenize(message)))
```

```python
def shared_vocabulary(first, second):
    return sorted(set(tokenize(first)) & set(tokenize(second)))

assert shared_vocabulary(*messages) == ["lab", "open"]
assert "not" in tokenize(messages[1])
assert tokenize("École 42") == ["école"]
print(shared_vocabulary(*messages))
print("Keep 'not': without it, the second message would lose its negation.")
```
